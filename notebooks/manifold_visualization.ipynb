## Configuration

```python
MIN_VISITS_LIST = (50, 75, 100)
REDUCTION_METHODS = ("tsne", "umap", "pca")
DICTIONARY_NAMES = ("neural", "behavioral", "paired")

MAX_VISITS = 151
SD = 42
BIN_SIZE = 0.3

ROUTE_MIN_VISITS = 100
ERROR_KEY = "raw_error"

print("Minimum visits:", MIN_VISITS_LIST)
print("Reduction methods:", REDUCTION_METHODS)
print("Dictionary types:", DICTIONARY_NAMES)
```

For each transition-dictionary representation, t-SNE, UMAP, and PCA embeddings are generated at minimum-visit thresholds of 50, 75, and 100.

- The predetermined actual route

## Standalone Neural-State Manifolds

Generate full-view neural-state manifolds for the min-visits=100 transition dictionary.

The actual predetermined route is compared with the highest-probability predicted route.

```python
SINGLE_DICTIONARY_TO_PLOT = "neural"
SINGLE_MIN_VISITS = 100

config = mv.get_dictionary_plot_config(SINGLE_DICTIONARY_TO_PLOT)

state_type = config["state_type"]
dict_type = config["dict_type"]

neural_state_dict, behavioral_state_dict, pair_transition_dict = mv.load_cache(min_visits=SINGLE_MIN_VISITS, max_visits=MAX_VISITS, sd=SD, bin_size=BIN_SIZE)
transition_dict = mv.select_transition_dict(SINGLE_DICTIONARY_TO_PLOT, neural_state_dict, behavioral_state_dict, pair_transition_dict)
matrix, state_keys = mv.convert_dict_to_matrix(transition_dict, dict_type=dict_type)

print("Matrix shape:", matrix.shape)
print("Number of states:", len(state_keys))
```

```python
route = np.load(os.path.join(PROJECT_ROOT, f"route_sequence_min{ROUTE_MIN_VISITS}.npy"), allow_pickle=True)
top3_route_history = np.load(os.path.join(PROJECT_ROOT, f"top3_route_transitions_min{ROUTE_MIN_VISITS}.npy"), allow_pickle=True).tolist()

print("Route transitions:", len(route))
print("Top-3 history frames:", len(top3_route_history))
```

```python
for method in REDUCTION_METHODS:
    method_label = mv.REDUCTION_LABELS[method]

    manifold = mv.perform_manifold_learning(matrix, method=method, n_components=3, random_state=42)

    mv.plot_manifold_raw_two_route(
        manifold,
        manifold_type=method_label,
        state_keys=state_keys,
        state_type=state_type,
        route_sequence=route,
        top3_route_history=top3_route_history,
        save_prefix=None,
        show_all_points=True,
        make_full_plot=True,
        make_zoomed_plot=False,
        zoom_padding_fraction=0.12,
        zoom_percentile=0,
        visual_offset_fraction=0.018,
        full_background_alpha=0.20,
        zoom_background_alpha=0.08,
        full_background_point_size=9,
        zoom_background_point_size=6
    )
```
